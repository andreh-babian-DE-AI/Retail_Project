```python
import pandas as pd
print("Jupyter is working1")

sales_data = pd.read_csv("retail_sales_clean_dataset.csv")
sales_data['date'] = pd.to_datetime(sales_data['date'])

print("csv loaded")

sales_by_date = sales_data.groupby('date')['sales'].sum().reset_index()
sales_by_region = sales_data.groupby('region')['sales'].sum().reset_index()
sales_by_product = sales_data.groupby('product')['sales'].sum().reset_index()
sales_by_price = sales_data.groupby('price')['sales'].sum().reset_index()

display(sales_data.head(20))
display(sales_by_date.head())
display(sales_by_region)



```

```python
import matplotlib.pyplot as plt

# Create a figure with a specific size
fig = plt.figure(figsize=(14, 11))  # Width: 10 inches, Height: 6 inches
fig.subplots_adjust(hspace=0.25, wspace=0.2)
# Create a figure with 2 rows and 2 columns of subplots
subplot1 = fig.add_subplot(2,2,1)
subplot2 = fig.add_subplot(2,2,2)
subplot3 = fig.add_subplot(2,2,3)
subplot4 = fig.add_subplot(2,2,4)


subplot1.bar(sales_by_region['region'],sales_by_region['sales'])  # Top-left subplot
subplot1.set_xlabel("Region")
subplot1.set_ylabel("total Sale")
subplot1.set_title("Sale by Region")
subplot1.legend()


subplot2.plot(sales_by_date['date'],sales_by_date['sales'])
subplot2.set_xlabel("date")
subplot2.set_ylabel("total Sale")

subplot3.pie(sales_by_product['sales'], labels = sales_by_product['product'], autopct ='%1.1f%%' )
subplot3.set_title("Products Sale Pie Chart ")

subplot4.scatter(sales_by_price['price'] , sales_by_price['sales'])
subplot4.set_xlabel("product price")
subplot4.set_ylabel("total Sale $")
subplot4.set_title("correlation between price and sale")


plt.show()
```
